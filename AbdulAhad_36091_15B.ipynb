Q3: Make a bar chart of snack choices

```python
import matplotlib.pyplot as plt
```

```python
snack_counts = tea["snack_choice"].value_counts()
snack_counts
```

```python
plt.bar(snack_counts.index, snack_counts.values, color="lightcoral", width=0.5)
plt.xlabel("Snack Choice")
plt.ylabel("Number of Students")
plt.title("Student Snack Choices")
plt.show()
```
